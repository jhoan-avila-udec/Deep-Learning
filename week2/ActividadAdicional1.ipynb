```python
import pandas as pd

class PerceptronBasico:
    def __init__(self, pesos, sesgo):
        self.pesos = pesos
        self.sesgo = sesgo

    def predecir(self, entradas):
        # 1. Cálculo del puntaje Z (Suma ponderada)
        # z = (x1 * w1 + x2 * w2 + ...) + b
        z = sum(x * w for x, w in zip(entradas, self.pesos)) + self.sesgo

        # 2. Decisión de salida (Función de activación Escalón)
        activacion = 1 if z >= 0 else 0

        return z, activacion

# --- CONFIGURACIÓN DE CONSTANTES ---
pesos_config = [0.3, -0.2]  # Pesos iniciales.
sesgo_config = 0.1          # Sesgo inicial.

modelo = PerceptronBasico(pesos_config, sesgo_config)

# --- CONJUNTO DE ENTRADAS (Casos de prueba) ---
# [MFCC, ZCR] Donde MFCC representan el timbre del sonido y ZCR indica la cantidad de ruido
entradas_casos_prueba = [
    [12.5, 0.18],
    [-5.3, 0.08],
    [15.1, 0.25],
    [-4, 0.05]
]
# TENGA EN CUENTA: Los rangos ZCR suelen ser (0.01 - 0.10) para acusticos.

# --- EJECUCIÓN Y REGISTRO DE RESULTADOS ---
resultados = []
for entradas in entradas_casos_prueba:
    z, salida = modelo.predecir(entradas)
    resultados.append({
        "Entrada (x1, x2)": entradas,
        "Puntaje Z": z,
        "Decisión (Salida)": salida,
        "Interpretación": "  Instrumento Electronico" if salida == 1 else " Instrumento Acustico"
    })

# --- MOSTRAR RESULTADOS ---
df_resultados = pd.DataFrame(resultados)
print("### Toma de decision entre sonido Acustico y Electronico con un Perceptrón ###\n")
print(df_resultados.to_string(index=False))
```

# **Explicación Red Neuronal**

Esta simple Red Neuronal busca comprender los conceptos mínimos de redes neuronales: entradas,
pesos, sesgo (bias), cálculo del puntaje z, regla de activación y obtención de una salida de
clasificación.

Por esta razon, plantee una neurona que pueda indicar como salida que tipo de instrumento musical (Acustico o Electronico) se le a proporcionado en la entrada.

Como parametros de entrada **x1 representa MFCC** y **x2 representa ZRC**. MFCC es el timbre que genera un instrumento y ZRC es la cantidad de ruido que genera.

Si los instrumentos son Acusticos los valores de las entradas seran bajos, pero si los valores son altos representara instrumentos Electronicos.
